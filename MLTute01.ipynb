LabelEncoder() function to automatically map all categories to numeric labels based on their alphabetical order.

```python
import pandas as pd
data_frame=pd.read_csv("/content/drive/MyDrive/ML/Week-01/Tute/Mall_Customers.csv")

data_frame.rename({'CustomerID':"ID", 'Gender':"Sex", 'Annual Income (k$)':"Salary"}, axis=1, inplace=True)

data_frame.head()
```

```python
data_frame['Sex'].unique()
```

Now let’s import the library sklearn, which has the preprocessing package with the LableEncoder() function to encode the categorical values for the variable “Sex”

```python
from sklearn import preprocessing
label_encoder = preprocessing.LabelEncoder()

label_encoder = preprocessing.LabelEncoder()
data_frame['Sex']= label_encoder.fit_transform(data_frame['Sex'])

data_frame.head()
```

```python
data_frame.rename({'Spending Score (1-100)':"Spending_Score"}, axis=1, inplace=True)
data_frame.head()
```
